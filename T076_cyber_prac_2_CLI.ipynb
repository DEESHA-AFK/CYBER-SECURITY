CLI

```python
import math

# Function to check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

# Extended Euclidean Algorithm
def extended_gcd(a, b):
    if a == 0:
        return (b, 0, 1)
    gcd, x1, y1 = extended_gcd(b % a, a)
    x = y1 - (b // a) * x1
    y = x1
    return (gcd, x, y)

# Modular Inverse
def mod_inverse(e, phi):
    gcd, x, y = extended_gcd(e, phi)
    if gcd != 1:
        return None
    return x % phi

# ---------------- MAIN PROGRAM ----------------

print("========== RSA Algorithm ==========\n")

# Input p
while True:
    p = int(input("Enter first prime number (p): "))
    if is_prime(p):
        break
    print("Error! p must be a prime number.\n")

# Input q
while True:
    q = int(input("Enter second prime number (q): "))
    if is_prime(q) and q != p:
        break
    print("Error! q must be a different prime number.\n")

# Calculate n and phi
n = p * q
phi = (p - 1) * (q - 1)

# Input e
while True:
    e = int(input(f"Enter value of e (1 < e < {phi}): "))
    if 1 < e < phi and math.gcd(e, phi) == 1:
        break
    print("Error! e must satisfy:")
    print("1 < e < phi(n)")
    print("gcd(e, phi) = 1\n")

# Calculate d
d = mod_inverse(e, phi)

print("\n----- Keys -----")
print("Public Key  = ({}, {})".format(e, n))
print("Private Key = ({}, {})".format(d, n))

# Plaintext
while True:
    message = int(input("\nEnter plaintext (number less than {}): ".format(n)))
    if 0 <= message < n:
        break
    print("Plaintext must be between 0 and", n - 1)

# Encryption
cipher = pow(message, e, n)

# Decryption
decrypted = pow(cipher, d, n)

print("\n----- Result -----")
print("Original Message :", message)
print("Encrypted Message:", cipher)
print("Decrypted Message:", decrypted)
```

GUI

```python
import math
import ipywidgets as widgets
from IPython.display import display

# Prime check
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

# Extended Euclidean Algorithm
def egcd(a, b):
    if a == 0:
        return b, 0, 1
    gcd, x1, y1 = egcd(b % a, a)
    x = y1 - (b // a) * x1
    y = x1
    return gcd, x, y

# Modular Inverse
def mod_inverse(e, phi):
    gcd, x, y = egcd(e, phi)
    if gcd != 1:
        return None
    return x % phi

# Input Widgets
p = widgets.IntText(value=17, description='Prime p:')
q = widgets.IntText(value=11, description='Prime q:')
e = widgets.IntText(value=7, description='e:')
msg = widgets.IntText(value=88, description='Message:')

button = widgets.Button(
    description="Encrypt & Decrypt",
    button_style='success'
)

output = widgets.Output()

# RSA Function
def rsa(btn):
    output.clear_output()

    with output:
        try:
            P = p.value
            Q = q.value
            E = e.value
            M = msg.value

            if not is_prime(P):
                print("❌ p must be a prime number.")
                return

            if not is_prime(Q):
                print("❌ q must be a prime number.")
                return

            if P == Q:
                print("❌ p and q must be different.")
                return

            n = P * Q
            phi = (P - 1) * (Q - 1)

            if E <= 1 or E >= phi or math.gcd(E, phi) != 1:
                print("❌ Invalid value of e.")
                return

            if M >= n:
                print(f"❌ Message must be less than {n}.")
                return

            d = mod_inverse(E, phi)

            cipher = pow(M, E, n)
            plain = pow(cipher, d, n)

            print("========== RSA RESULT ==========")
            print(f"Public Key  : ({E}, {n})")
            print(f"Private Key : ({d}, {n})")
            print(f"Original Message : {M}")
            print(f"Encrypted Message: {cipher}")
            print(f"Decrypted Message: {plain}")

        except:
            print("❌ Invalid Input!")

button.on_click(rsa)

display(
    widgets.VBox([
        widgets.HTML("<h2>🔐 RSA Encryption & Decryption</h2>"),
        p,
        q,
        e,
        msg,
        button,
        output
    ])
)
```
